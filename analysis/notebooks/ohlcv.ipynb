Imports

```python
import sqlite3
from pathlib import Path

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
```

Database Connection

```python
ROOT = Path.cwd().parents[1]

DB_PATH = ROOT / "storage" / "database" / "ETH.db"

print(DB_PATH)

conn = sqlite3.connect(DB_PATH)
```

Load Data

```python
query = """
SELECT *
FROM ohlcv
ORDER BY open_time
"""

df = pd.read_sql(query, conn)

df.head()
```

Dataset Information

```python
print(df.shape)

df.info()
```

Convert Timestamp

```python
df["open_datetime"] = pd.to_datetime(
    df["open_time"],
    unit="ms",
    utc=True
)

df["close_datetime"] = pd.to_datetime(
    df["close_time"],
    unit="ms",
    utc=True
)

df.head()
```

Set Index

```python
df.set_index("open_datetime", inplace=True)

df.head()
```
